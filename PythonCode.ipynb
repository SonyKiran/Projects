```python
import pandas as pd
import smtplib
from email.mime.multipart import MIMEMultipart
from email.mime.text import MIMEText
from email.mime.application import MIMEApplication
import time
import os

# Configuration
EXCEL_FILE = "recruiters.xlsx"   # Your Excel file with recruiter emails
RESUME_FILE = "Kiran_Devarapalli.pdf"  # Your resume file
EMAIL_SUBJECT = "Job Application - Sony Kiran"

# Your email credentials
YOUR_EMAIL = "kirandevarap@gmail.com"
YOUR_PASSWORD = "peoj pfcj jkhe cjxm"  # Use Gmail App Password
SMTP_SERVER = "smtp.gmail.com"
SMTP_PORT = 587

def send_emails():
    # Read all sheets from Excel file
    excel_file = pd.ExcelFile(EXCEL_FILE)
    all_emails = []

    # Process each sheet
    for sheet_name in excel_file.sheet_names:
        df = pd.read_excel(excel_file, sheet_name=sheet_name)

        # Check if 'mail' column exists
        if 'mail' in df.columns:
            valid_emails = df['mail'].dropna().tolist()
            all_emails.extend(valid_emails)
        else:
            print(f"Warning: No 'mail' column in sheet '{sheet_name}'")

    # Remove duplicates
    all_emails = list(set(all_emails))
    print(f"Found {len(all_emails)} unique email addresses.")

    # Connect to SMTP server
    server = smtplib.SMTP(SMTP_SERVER, SMTP_PORT)
    server.starttls()
    server.login(YOUR_EMAIL, YOUR_PASSWORD)

    # Send emails
    for recipient in all_emails:
        msg = MIMEMultipart()
        msg['From'] = YOUR_EMAIL
        msg['To'] = recipient
        msg['Subject'] = EMAIL_SUBJECT

        # Email body
        body = f"""
Dear Recruiter,

I am writing to express my interest in Business Analyst opportunities at your Esteemed company.
Please find my resume attached for your consideration.

I look forward to discussing how my skills and experience could benefit your team.

Best regards,
Sony
Phone: 408-590-3037
"""
        msg.attach(MIMEText(body, 'plain'))

        # Attach resume
        try:
            with open(RESUME_FILE, 'rb') as file:
                attachment = MIMEApplication(file.read(), _subtype="pdf")
                attachment.add_header(
                    'Content-Disposition',
                    'attachment',
                    filename=os.path.basename(RESUME_FILE)
                )
                msg.attach(attachment)
        except Exception as e:
            print(f"Error attaching resume: {e}")
            continue

        # Send email
        try:
            server.send_message(msg)
            print(f"Email sent to {recipient}")
            time.sleep(3)  # delay to avoid spam/rate limits
        except Exception as e:
            print(f"Failed to send email to {recipient}. Error: {e}")

    server.quit()
    print("All emails processed!")

if __name__ == "__main__":
    print("Sending emails now...")
    send_emails()
```
